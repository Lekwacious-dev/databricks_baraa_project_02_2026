# Transformation Logic

```python
query = """
SELECT 
    ROW_NUMBER() OVER (ORDER BY p.start_date, p.product_key) AS product_key, -- Surrogate key
    p.product_id,
    p.product_key AS product_number,
    p.product_name,
    p.product_cost,
    p.product_line,
    p.start_date,
    p.end_date,
    p.category_id,
    pc.category,
    pc.subcategory,
    pc.maintenance
FROM silver.crm_products p
LEFT JOIN silver.erp_product_category pc
ON pc.category_id = p.category_id
"""
df = spark.sql(query)
```

```python
df.limit(10).display()
```

# Write to Gold Table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.dim_products")
```
