# Transformation Logic

```python

query = """

SELECT
  ROW_NUMBER() OVER (ORDER BY c.customer_id, c.customer_number) 
  AS customer_key,
  c.customer_id,
  c.customer_number,
  c.customer_firstname,
  c.customer_lastname,
  c.marital_status,
  COALESCE(e.gender, c.customer_gender) AS gender,
  e.birth_date,
  l.country
FROM silver.crm_customers c
LEFT JOIN silver.erp_customers e
  ON c.customer_number = e.customer_id
LEFT JOIN silver.erp_customer_location l
  ON c.customer_number = l.customer_id
"""


df = spark.sql(query)
```

```python
display(df)
```

# Writing to Gold Table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.dim_customers")
```
