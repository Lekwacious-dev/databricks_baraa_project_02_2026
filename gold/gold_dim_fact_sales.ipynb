#Transformation Logic

```python
query = """
SELECT 
    sd.order_id,
    pr.product_number,
    cu.customer_key,
    sd.order_date,
    sd.ship_date,
    sd.due_date,
    sd.sales_amount,
    sd.quantity,
    sd.price
FROM silver.crm_sales sd
LEFT JOIN gold.dim_products pr
    ON sd.product_number = CAST(pr.product_key AS STRING)
LEFT JOIN gold.dim_customers cu
    ON sd.customer_id = cu.customer_id
"""
df = spark.sql(query)
```

```python
df.limit(10).display()
```

# Write to Gold Table

```python
df.write.mode("overwrite").format("delta").option("mergeSchema", "true").saveAsTable("workspace.gold.dim_fact_sales")
```
